```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# --- Data ---
area = np.array([50, 65, 70, 85, 100, 120, 140, 160, 180, 200])
price = np.array([15000, 19500, 21000, 25500, 30000, 36000, 42000, 48000, 54000, 60000])

n = len(area)

# --- Step 1: Manual Gradient Descent ---
def gradient_descent(area, price, alpha, iterations=1000):
    theta0, theta1 = 0.0, 0.0
    n = len(area)
    cost_history = []

    for i in range(iterations):
        y_pred = theta0 + theta1 * area
        error = y_pred - price

        cost = (1 / (2 * n)) * np.sum(error ** 2)
        cost_history.append(cost)

        d_theta0 = (1 / n) * np.sum(error)
        d_theta1 = (1 / n) * np.sum(error * area)

        theta0 = theta0 - alpha * d_theta0
        theta1 = theta1 - alpha * d_theta1

    return theta0, theta1, cost_history

# Good learning rate
theta0, theta1, cost_history = gradient_descent(area, price, alpha=0.0001, iterations=1000)

print("Manual Gradient Descent Result:")
print(f"theta0: {theta0:.4f}")
print(f"theta1: {theta1:.4f}")

# --- Step 2: Compare to sklearn ---
X = area.reshape(-1, 1)
model = LinearRegression()
model.fit(X, price)

print("\nsklearn Result:")
print(f"theta0 (intercept): {model.intercept_:.4f}")
print(f"theta1 (coef): {model.coef_[0]:.4f}")

# --- Step 3: Try a large alpha (diverges) ---
theta0_bad, theta1_bad, cost_history_bad = gradient_descent(area, price, alpha=0.01, iterations=20)

print("\nWith alpha too large (0.01), first 10 costs:")
print(cost_history_bad[:10])

# --- Step 4: Plot cost vs iterations (good alpha) ---
plt.figure(figsize=(8, 6))
plt.plot(range(len(cost_history)), cost_history, color='blue')
plt.xlabel('Iteration')
plt.ylabel('Cost (J)')
plt.title('Cost Function vs Iterations (alpha=0.0001)')
plt.grid(True, alpha=0.3)
plt.show()
# plt.savefig('/mnt/user-data/outputs/cost_convergence.png', dpi=150, bbox_inches='tight')
plt.close()
```
